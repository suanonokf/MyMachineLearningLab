```python
import pandas as pd
url ="https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
columns = ['sepal_length','sepal_width','petal_length','petal_width','species']
csv =pd.read_csv(url)
df=pd.DataFrame(csv.values,columns=columns)
print("Iris species of type Iris-setosa")
print(df[df['species']=='Iris-setosa'].head())
print("Iris with a sepal lenght less than 4.5")
print(df[df['sepal_length']<4.5])

print("Multiply the column sepal lenght values by 2")
df['sepal_length']=df['sepal_length'].astype(int)
mapped_df = df.apply(lambda x: x*2)
print(mapped_df.head())
print("\n Missing values handles")
df_copy = df.copy()
print(df_copy.dropna(subset=['sepal_length']))
saved=df.to_csv("saved.csv")
print("\n Saved saved file")
print(saved)
print()
```

```python
import pandas as pd
import numpy as np

# Create sample data
data = {
'Department': ['Engineering', 'Marketing', 'Engineering', 'Sales',
'Marketing', 'Engineering', 'Sales', 'Marketing'],
'Employee': ['Alice', 'Bob', 'Charlie', 'David',
'Eve', 'Frank', 'Grace', 'Henry'],
'Salary': [70000, 65000, 80000, 60000, 68000, 75000, 62000, 70000],
'Years': [2, 5, 8, 3, 6, 4, 2, 7]
}
df = pd.DataFrame(data)
print("Original Data Frame")
print(df)
print()
print("Group By department")
print(df.groupby('Department').describe())
print()
print("Aggregation Operations")
group_by_years=df.groupby('Years')

print("Minimum Year Value")
print(group_by_years.agg('min'))

print("\nMaximum Year Value")
print(group_by_years.agg('max'))

print("\nList of aggregate operations  columns")
print(group_by_years.agg(['min','max']))

print("\nMultiple aggregations for multiple columns")
agg_dict={
    'Department':['min','max'],
    'Salary':['sum','mean'],
}
print(group_by_years.agg(agg_dict))
df['Experience'] = pd.cut(
    df['Years'],
    bins=[0,3,6,10],
    labels=['junior','Mid','senior']
)
print("\n",df)
```

```python
import pandas as pd
employees = pd.DataFrame({
    'EmployeeID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'DeptID': [101, 102, 101, 103]
})

departments = pd.DataFrame({
    'DeptID': [101, 102, 103],
    'Department': ['Engineering', 'Marketing', 'Sales']
})
# merging dataframes
inner_merge = pd.merge(employees,departments,on='DeptID')
print("\nInner Join")
print(inner_merge)
#left Join
new_employee = pd.DataFrame(
    {
        'EmployeeID':[5],
        'Name':['Greg'],
        'DeptID':[105]
    }
)
all_employees = pd.concat([employees,new_employee],ignore_index=True)
merged_left = pd.merge(all_employees,departments,how='right',on='DeptID')
print(merged_left)
```
