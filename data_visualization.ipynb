```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.core.pylabtools import figsize
from matplotlib.pyplot import figure

x = np.linspace(0,3*np.pi,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10,6))
plt.plot(x,y1,label='sin(x)',linewidth=2)
plt.plot(x,y2,label='cos(x)',linewidth=2,linestyle='--')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Trigonometric function')
plt.legend()
plt.grid(True,alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = x * np.exp(-x/10)
y3 = np.cos(x)

plt.figure(figsize=(12,12))
plt.plot(x,y1,'b-',label='sin(x)',linewidth=2,)
plt.plot(x,y2,'r-',label='damped sin(x)',linewidth=2)
plt.plot(x,y3,'g-',label='cos(x)',linewidth=2)

plt.xlabel('Time',fontsize=10)
plt.ylabel('Amplitude',fontsize=10)
plt.title('Oscillating function', fontsize=10, fontweight='bold')
plt.legend()
plt.grid(True,alpha=0.3)
plt.axhline(0,linestyle='--',color='k')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,5,10)
y = x**2
plt.figure(figsize=(10,6))

plt.plot(x,y,marker='o',label='y=x^2',linewidth=2,
         markersize=5,color='purple',markerfacecolor='yellow',markeredgecolor='black')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True,alpha=0.3)
plt.title('Square Function', fontsize=10, fontweight='bold')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
fig, axes=plt.subplots(2,2,figsize=(12,12))
x=np.linspace(0,10,100)
axes[0,0].plot(x,x,'b-',label ='y=x',linewidth=2)
axes[0,0].set_title('linear function')
axes[0,0].legend()

axes[0,1].plot(x,x**2,'r-',label='y=x^2',linewidth=2)
axes[0,1].set_title('quadratic function')
axes[0,1].legend()

axes[1,0].plot(x,np.sqrt(x),'g-',label ='y=sqrt(x)',linewidth=2)
axes[1,0].set_title('square-root function')
axes[1,0].legend()
axes[1,1].plot(x,x**3,'y-',label='y=x^3',linewidth=2)
axes[1,1].set_title('cube function')
axes[1,1].legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,10,100)
y = np.sin(x) * np.exp(-x/10)
plt.figure(figsize=(10,6))
plt.plot(x,y,'b-',label='damped sin(x)',linewidth=2)

#Add peaks
peaks = x[np.where((y[1:-1]>y[0:-2]) & (y[1:-1]>y[2:]))]
for peak in peaks[:3]:
    plt.axvline(peak,linestyle='--',color='k')
# Customize appearance
plt.xlabel('x', fontsize=14, fontweight='bold')
plt.ylabel('y', fontsize=14, fontweight='bold')
plt.title('Damped Sine Wave with Annotations', fontsize=16)
plt.legend(loc='upper right', fontsize=12)
plt.grid(True, alpha=0.3, linestyle=':')
plt.xlim(0, 10)
plt.show()
```
