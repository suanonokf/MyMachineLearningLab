```python
from cProfile import label

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
x= np.random.randn(100)
y = x*2 + np.random.randn(100)*0.5

# Scatter plot
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.scatter(x,y,s=50,alpha=0.5,label='data')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title("Scatter plot")
plt.axhline(y=0, color='r')
plt.show()

#Histogram
plt.figure(figsize=(10,6))
plt.subplot(1,2,2)
plt.hist(y,bins=20,alpha=0.6,edgecolor='white',label='data')
plt.xlabel('value')
plt.ylabel('frequency')
plt.title('Histogram of data')
plt.grid(True,alpha=0.3)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

n=200
categories = np.random.choice(['A','B','C'],n)
x=np.random.randn(n)
y=np.random.randn(n)

sizes = np.random.randint(20,200,n)
colors ={
    'A':'blue',
    'B':'red',
    'C':'green'
}
plt.figure(figsize=(10,8))
for category in ['A','B','C']:
    mask= categories==category
    plt.scatter(
        x[mask],
        y[mask],
        s=sizes[mask],
        label=f"category {category}",
        linewidths=0.5
    )
plt.xlabel('x',fontsize=12)
plt.ylabel('y',fontsize=12)
plt.title('Scatter plot with categorical data',fontsize=12,fontweight='bold')
plt.legend()
plt.show()
```
