```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SimpleConv2D(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels=4, out_channels=32, kernel_size=3, stride=1,)
        self.MaxPool1 = nn.MaxPool2d(2,2)

        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1,)
        self.MaxPool2 = nn.MaxPool2d(2,2)

        self.conv3 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3, stride=1,)
        self.MaxPool3 = nn.MaxPool2d(2,2)

        self.fc1 = nn.Linear(16*16*128,256)
        self.fc2 = nn.Linear(256,64)
        self.fc3 = nn.Linear(64,1)
    def forward(self,x):
        #First block
        x = self.conv1(x)
        x = F.relu(x)
        x =self.MaxPool1(x)
        print("Conv1 + MaxPool1: ",x.shape)
        #Sceond Block
        x = self.conv2(x)
        x = F.relu(x)
        x =self.MaxPool2(x)
        print("Conv2 + MaxPool2: ",x.shape)

        x = self.conv3(x)
        x = F.relu(x)
        x =self.MaxPool3(x)
        print("Conv2 + MaxPool3: ",x.shape)

        #Classifiers
        x = x.flatten(start_dim=1)
        print(f"Flatten Shape: {x.shape}")

        x = self.fc1(x)
        x= F.relu(x)
        x = self.fc2(x)
        x = F.relu(x)
        x = self.fc3(x)

        return x


```
