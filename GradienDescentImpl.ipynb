```python
from unittest.mock import PropertyMock

import numpy as np
import pandas as pd


class GradientDescentImpl:
    def __init__(self,learning_rate=0.1,iterations=400):
        self.learning_rate = learning_rate
        self.iterations = iterations
        self.weight = np.zeros(shape=(5,1))
        self.bias = 0
        self.costHistory = []

    def fit(self,X,y):

        m=len(y)
        for i in range(self.iterations):
            y_pred = np.dot(X,self.weight)+ self.bias

            dw = (1/m)*np.dot(X.T,(y_pred - y))
            db = (1/m)*np.sum(y_pred-y)

            self.weight = self.weight-self.learning_rate*dw
            self.bias = self.bias-self.learning_rate*db

            cost = (1/(2*m))*np.sum((y_pred-y)**2)
            self.costHistory.append(cost)

            if i==399:
                return self.weight,self.bias

    def predict(self,X):
        return np.dot(self.weight.T,X.T) + self.bias

    def set_learning_rate(self):
        self.learning_rate = self.learning_rate
    def set_iterations(self):
        self.iterations = self.iterations
    def set_weight(self):
        self.weight = self.weight
    def set_bias(self):
        self.bias = self.bias

    @property
    def get_learning_rate(self):
        return self.learning_rate
    @property
    def get_iterations(self):
        return self.iterations
    @property
    def get_weight(self):
        return self.weight
    @property
    def get_bias(self):
        return self.bias

def standardize(X):
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0)
    return (X - mean) / std, mean, std

df =pd.read_csv('Real-estate1.csv')
df_copy = df.copy()
dictionnary = df_copy.to_dict()
#Retrieve all features
house_age = np.array(list(dictionnary['X2 house age'].values()))
distance_to_the_nearest_station = np.array(list(dictionnary['X3 distance to the nearest MRT station'].values()))
number_of_convenient_stores = np.array(list(dictionnary['X4 number of convenience stores'].values()))
latitude = np.array(list(dictionnary['X5 latitude'].values()))
longitude = np.array(list(dictionnary['X6 longitude'].values()))
price = np.array([list(dictionnary['Y house price of unit area'].values())])
#Subdivide the dataset into training et predicting sets // total number of rows is 415 (0-414)
#Training sets
t_house_age = house_age[:400]
t_distance_to_the_nearest_station = distance_to_the_nearest_station[:400]
t_number_of_convenient_stores = number_of_convenient_stores[:400]
t_latitude = latitude[:400]
t_longitude = longitude[:400]
t_price = price[:,:400]

standardize_house_age =standardize(t_house_age)
standardize_distance = standardize(t_distance_to_the_nearest_station)
standardize_number_of_convenient_stores = standardize(t_number_of_convenient_stores)
standardize_latitude = standardize(t_latitude)
standardize_longitude = standardize(t_longitude)
test_X = np.array([
    standardize_house_age[0],
    standardize_distance[0],
    standardize_number_of_convenient_stores[0],
    standardize_latitude[0],
    standardize_longitude[0],
]).T
#Predicting sets
p_house_age = house_age[400:]
p_distance_to_the_nearest_station = distance_to_the_nearest_station[400:]
p_number_of_convenient_stores = number_of_convenient_stores[400:]
p_latitude = latitude[400:]
p_longitude = longitude[400:]

p_price = price[:,400:]
p_stand_house_age = standardize(p_house_age)
p_stand_distance = standardize(p_distance_to_the_nearest_station)
p_stand_number_of_convenient_stores = standardize(p_number_of_convenient_stores)
p_stand_latitude = standardize(p_latitude)
p_stand_longitude = standardize(p_longitude)
predicting_X = np.array(
    [
        p_stand_house_age[0],
        p_stand_distance[0],
        p_stand_number_of_convenient_stores[0],
        p_stand_latitude[0],
        p_stand_longitude[0],
    ]
).T
#Train the model
gradientDescent = GradientDescentImpl()
gradientDescent.fit(test_X,t_price.T)
predict_cost_of_house=gradientDescent.predict(predicting_X)
print("Predicted Cost:",predict_cost_of_house,"\n")
print("Actual Cost of house",list(p_price),"\n")
```
