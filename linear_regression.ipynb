```python
#Import packages
#Read the csv file
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error
from sklearn import preprocessing

df = pd.read_csv('Real-estate1.csv')
df.drop('No',axis=1,inplace=True)
print(df.head())
```

```python
#Draw a scatter plot between the number of convenience stores and the price houses
import matplotlib.pyplot as plt
number_convenience_stores = df['X4 number of convenience stores']
house_price = df['Y house price of unit area']
plt.figure(figsize = (10,10))
plt.scatter(number_convenience_stores,house_price,label='data',color='green')
plt.xlabel('Number of convenience stores')
plt.ylabel('House price')
plt.legend()
plt.title('Scatter plot of number of convenience stores according to house price',fontsize = 20,fontweight='bold')
plt.show()
```

```python
#Split the dependant variable which is the price of the house with the independant variables
Y = df['Y house price of unit area']
X = df.drop('Y house price of unit area',axis=1)

#Split the dataset into training and testing sets using the train_test_split package
X_train,X_test,Y_train,Y_test=train_test_split(X, Y, test_size=0.3, random_state=101)
#Cerate a Linear regression model
model = LinearRegression()
#Fit the model
model.fit(X_train,Y_train)
print("Coefficients: ",model.coef_)
print("Intercept: ",model.intercept_)
```

```python
#make a prdeiction with the the test set
y_pred = model.predict(X_test)
mse = mean_squared_error(Y_test,y_pred)
print("Mean Squared Error =",mse)
mae = mean_absolute_error(Y_test,y_pred)
print("Mean Absolute Error =",mae)
```
